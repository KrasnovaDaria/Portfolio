# 2. Гипотезы

Для дальнейших исследований добавим колонки:

- is_prom - индикатор применения промокода (1 если применялся, 0 если не применялся)

- is_complete - индикатор статуса траннзакии (1 если выполнена, 0 если нет)

- revenue - стоимость транзакии после скидки

- month - месяц транзакии

- week - неделя транзакции

- dow - день недели транзакции (0 - пн, 6 - вс)

```python
df['is_prom'] = df['promocode']
df['is_prom'] = df['is_prom'].fillna(0)
df.loc[df['is_prom'] != 0, 'is_prom'] = 1
```

```python
df['is_complete'] = df['order_state']
df.loc[df['order_state'] == 'complete', 'is_complete'] = 1
df.loc[df['order_state'] != 'complete', 'is_complete'] = 0
```

```python
df['month'] = df['order_date'].dt.month
```

```python
df['week'] = df['order_date'].dt.isocalendar().week
```

```python
df['dow'] = df['order_date'].dt.weekday # 0-пн, 6 - вс
```

```python
df['revenue'] = df['price'] - df['discount']
df.sample(10)
```

Имеет смысл рассматривать отдельно проведенные и непроведенные транзакции, поэтому разделим датасет на два

```python
df_com = df[df['is_complete']== 1]
df_incom = df[df['is_complete']== 0]
```

```python
df_corr = df[['order_date', 'price', 'discount', 'discount_percent', 'is_prom', 'is_complete', 'month', 'week', 'dow', 'revenue']]
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df_corr.corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Корреляционная матрица')
plt.show()
```

Возможно есть линейная зависимость между статусом трназакции и применением промокода, а так же между ценой и скидкой.

### 2.3.4 Зависисмость количества и суммы успешных транзакций от применения прмокода

```python
plt.figure(figsize=(18, 5))

ax1 = plt.subplot(1, 2, 1)

df_com\
.groupby('order_date')\
.agg({'revenue' : 'sum', 'is_prom' : 'mean'})\
.reset_index().plot(\
                    y='is_prom', \
                    x='revenue', \
                    kind='scatter', \
                    grid=True, ax=ax1, \
                    xlabel='сумма транзакций (млн.р.)', \
                    ylabel='% транзакций с применением промокода', \
                    title='Зависимость суммы продаж от применения промокодов')

ax2 = plt.subplot(1, 2, 2)

df_com\
.groupby('order_date')\
.agg({'revenue' : 'count', 'is_prom' : 'mean'})\
.reset_index().plot(\
                    y='is_prom', \
                    x='revenue', \
                    kind='scatter', \
                    grid=True, ax=ax2, \
                    xlabel='количество транзакций', \
                    ylabel='% транзакций с применением промокода', \
                    title='Зависимость количества продаж от применения промокодов')
```

### 2.3.5. Зависимость количества и суммы неуспешных транзакций от применения промокода

```python
plt.figure(figsize=(18, 5))

ax1 = plt.subplot(1, 2, 1)

df_incom\
.groupby('order_date')\
.agg({'revenue' : 'sum', 'is_prom' : 'mean'})\
.reset_index().plot(\
                    y='is_prom', \
                    x='revenue', \
                    kind='scatter', \
                    grid=True, ax=ax1, \
                    xlabel='сумма транзакций (млн.р.)', \
                    ylabel='% транзакций с применением промокода', \
                    title='Зависимость суммы продаж от применения промокодов')

ax2 = plt.subplot(1, 2, 2)

df_incom\
.groupby('order_date')\
.agg({'revenue' : 'count', 'is_prom' : 'mean'})\
.reset_index().plot(\
                    y='is_prom', \
                    x='revenue', \
                    kind='scatter', \
                    grid=True, ax=ax2, \
                    xlabel='количество транзакций', \
                    ylabel='% транзакций с применением промокода', \
                    title='Зависимость количества продаж от применения промокодов')
```

```python
def visual(df, title_1, title_2, title_3, xlabel, ylabel):
    print (title_1)

    display (df.describe())

    plt.figure(figsize=(17.75,3))
    sns.boxplot(
        df,
        orient='h'
    ).set(
        title = title_2,
        xlabel = xlabel,
        ylabel = '')
    plt.show()

    plt.figure(figsize=(17,5))
    sns.histplot(
        df,
        bins=100
    ).set(title = title_3,
        xlabel = xlabel,
        ylabel = ylabel)
    plt.show()
```
